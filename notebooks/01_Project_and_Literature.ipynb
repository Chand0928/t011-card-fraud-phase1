# 01. Project Overview and Literature Review
**Team T011, Section A, Project 26. Phase 1: Machine Learning.**

## What are we building?
A program that puts unusual card transactions into a queue for human review. It learns legitimate patterns using a Gaussian Mixture Model (GMM). Larger anomaly scores mean less typical transactions. An alert does not prove fraud.

This is a reading notebook. It has no training code. Read it first, then run notebooks 02, 03, 04 and 05 in order.

## Our question
Can a GMM learned from earlier legitimate transactions help rank later fraud for review?

## What does the dataset contain?
One row is one transaction. Time is elapsed seconds, V1-V28 are anonymous provider PCA features, Amount is in EUR, and Class is the correct label: 0 legitimate, 1 fraud. The provider used PCA to replace original variables with numerical combinations. We do not run another PCA.

The raw file has 284,807 rows and 492 frauds. Our documented exact-repeat removal leaves 283,726 rows and 473 frauds. The actual quality checks happen in notebook 02.

## Literature review: what each source contributes
| Source | Main idea in simple words | Our use | Limit to explain |
|---|---|---|---|
| Dempster, Laird and Rubin (1977) | EM alternates estimated hidden memberships with parameter updates | Explain how the GMM fits components | Convergence does not prove the globally best solution |
| Saito and Rehmsmeier (2015) | Precision-recall measures expose false-alert burden when positives are rare | Use PR AUC, recall and review precision rather than headline accuracy | PR performance depends on the test base rate |
| scikit-learn mixture guide | GMM covariance settings and information criteria give a usable baseline | Diagonal covariance; report AIC and BIC; choose minimum training BIC | Better legitimate density fit does not guarantee better fraud detection |
| ULB/Worldline data and TensorFlow mirror documentation | Provide the recommended benchmark and a documented download path | Real transaction data with a recorded source and checksum | Old, short history and anonymous inputs limit deployment claims |

**Why this is a literature review:** We connect each source to a decision and explain its limitation. We are not claiming our method is a new algorithm or copying another paper's performance numbers.

The assignment also recommends Sakurada and Yairi (2014), *Anomaly Detection Using Autoencoders*. That reading belongs to Phase 2 and supplies no Phase 1 results here.

## Our evaluation plan, fixed before test results
- Use earlier training rows, a later validation period and the latest test period.
- Fit the scaler and GMM only on legitimate training rows.
- Try K=1,2,4,8 and select the smallest training BIC among converged candidates.
- Set alert thresholds on legitimate validation scores at 0.1%, 0.5% and 1% FPR budgets. The illustrative default is 0.5%.
- Evaluate the frozen choices on mixed test data. Report trapezoidal PR AUC, recall at the validation policies, actual test FPR and precision at fixed review capacities.
- Compare the selected GMM with the K=1 candidate under the same data, scaler and scoring rule. This reference adds no new model family or search.
- Keep a record of assumptions, convergence, inputs and measured results.

An unsupervised GMM does not receive target labels during fitting, but we do use known labels to select legitimate training examples. This is legitimate-only novelty detection, not completely label-free preparation.

## What is already measured?
The default policy finds 37 of 74 test frauds, misses 37, and creates 172 legitimate false alerts. Precision@100 is 29%. These are previous measured results of the same fixed experiment. Notebook 05 must reproduce them; we do not tune against them.
